```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.express as px

#Load gapminder data
df = px.data.gapminder()
algeria_df=df[df['country']=='Algeria']

#Subsets
africa_1952 = df[(df['continent']=='Africa')&(df['year']==1952)].sort_values(by='pop')
# algeria_df
africa_1952

plot=px.bar(x='country',y='pop',data_frame=africa_1952)
plot.update_layout(title='Population of Africa in 1952')
plot.show()
```

```python
africa_2007 = df[(df['continent']=='Africa')&(df['year']==2007)].sort_values(by='pop')

plot=px.bar(x='country',y='pop',data_frame=africa_2007)
plot.update_layout(title='Population of Africa in 2007')
plot.show()
```

```python
africa_2007
```

```python
goplot=go.Figure(px.scatter(africa_2007,x='gdpPercap',y='lifeExp',color='country' ))
goplot.update_layout(title='GDP vs Life Expectancy')

goplot.show()
```
