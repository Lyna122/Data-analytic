```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots


# Seattle weather data - COMPLETE 12 MONTHS OF DATA
seattle_weather_data = [
    ["USW00094290", "2014-01-01", 4.81, 42.1, 3.79, 5.88],
    ["USW00094290", "2014-02-01", 3.31, 43.4, 2.52, 4.28],
    ["USW00094290", "2014-03-01", 3.51, 46.6, 2.62, 4.42],
    ["USW00094290", "2014-04-01", 2.77, 50.5, 1.88, 3.78],
    ["USW00094290", "2014-05-01", 2.16, 56.0, 1.35, 3.15],
    ["USW00094290", "2014-06-01", 1.63, 61.0, 0.92, 2.47],
    ["USW00094290", "2014-07-01", 0.79, 65.9, 0.39, 1.35],
    ["USW00094290", "2014-08-01", 0.97, 66.5, 0.48, 1.63],
    ["USW00094290", "2014-09-01", 1.52, 61.6, 0.75, 2.48],
    ["USW00094290", "2014-10-01", 3.41, 53.3, 2.24, 4.82],
    ["USW00094290", "2014-11-01", 5.84, 46.2, 4.55, 7.28],
    ["USW00094290", "2014-12-01", 5.43, 41.1, 4.21, 6.82]
]
seattle_weather = pd.DataFrame(seattle_weather_data,
                               columns=['STATION', 'DATE', 'MLY-PRCP-NORMAL',
                                       'MLY-TAVG-NORMAL', 'MLY-PRCP-25PCTL', 'MLY-PRCP-75PCTL'])
seattle_weather = seattle_weather[seattle_weather["STATION"] == "USW00094290"]
seattle_weather.set_index("DATE", inplace=True)

# Austin weather data - COMPLETE 12 MONTHS OF DATA
austin_weather_data = [
    ["2014-01-01", 2.20, 49.2, 1.18, 3.35],
    ["2014-02-01", 2.31, 52.1, 1.25, 3.52],
    ["2014-03-01", 2.78, 59.2, 1.65, 4.12],
    ["2014-04-01", 2.27, 67.3, 1.22, 3.48],
    ["2014-05-01", 4.05, 75.2, 2.85, 5.52],
    ["2014-06-01", 4.63, 81.0, 3.28, 6.25],
    ["2014-07-01", 2.44, 83.4, 1.35, 3.78],
    ["2014-08-01", 2.03, 83.7, 1.08, 3.22],
    ["2014-09-01", 2.54, 77.8, 1.48, 3.88],
    ["2014-10-01", 4.24, 69.0, 2.95, 5.82],
    ["2014-11-01", 2.94, 58.4, 1.82, 4.35],
    ["2014-12-01", 2.21, 50.4, 1.22, 3.42]
]
austin_weather = pd.DataFrame(austin_weather_data,
                              columns=['DATE', 'MLY-PRCP-NORMAL', 'MLY-TAVG-NORMAL',
                                      'MLY-PRCP-25PCTL', 'MLY-PRCP-75PCTL'])
austin_weather.set_index("DATE", inplace=True)

# Adding the month column - FIXED VERSION
month = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
seattle_weather["MONTH"] = month
austin_weather["MONTH"] = month

print("Data loaded successfully!")
print(f"\nSeattle Weather Shape: {seattle_weather.shape}")
print(f"Austin Weather Shape: {austin_weather.shape}")
print("\nSeattle Weather Preview:")
print(seattle_weather.head())


print("="*70)

fig1 = go.Figure()

fig1.add_traces(px.line(seattle_weather,
                       x='MONTH',
                       y='MLY-PRCP-NORMAL',
                      ).update_traces(mode='lines',
                                      line=dict(color='red'),
                                      name='Seattle').data
              )

fig1.add_traces(px.line(austin_weather,
                        x='MONTH',
                        y='MLY-PRCP-NORMAL',
                       ).update_traces(mode='lines',
                                       line=dict(color='blue'),
                                       name='Austin').data
               )

fig1.update_layout(
    title="Austin vs Seattle Average Rainfall",
    xaxis_title="Month",
    yaxis_title="Precipitation (inches)",
    showlegend=True
)

fig1.show()


print("="*70)

# Create a subplot grid with 2 rows and 2 columns
fig2 = make_subplots(rows=2, cols=2,
                     subplot_titles=("Seattle Precipitation", "Seattle Temperature",
                                     "Austin Precipitation", "Austin Temperature"))

# Add Seattle precipitation to the top-left (row=1, col=1)
fig2.add_trace(
    go.Scatter(x=seattle_weather["MONTH"], y=seattle_weather["MLY-PRCP-NORMAL"],
               mode='lines', name="Seattle Precipitation"),
    row=1, col=1
)

# Add Seattle temperature to the top-right (row=1, col=2)
fig2.add_trace(
    go.Scatter(x=seattle_weather["MONTH"], y=seattle_weather["MLY-TAVG-NORMAL"],
               mode='lines', name="Seattle Temperature"),
    row=1, col=2
)

# Add Austin precipitation to the bottom-left (row=2, col=1)
fig2.add_trace(
    go.Scatter(x=austin_weather["MONTH"], y=austin_weather["MLY-PRCP-NORMAL"],
               mode='lines', name="Austin Precipitation"),
    row=2, col=1
)

# Add Austin temperature to the bottom-right (row=2, col=2)
fig2.add_trace(
    go.Scatter(x=austin_weather["MONTH"], y=austin_weather["MLY-TAVG-NORMAL"],
               mode='lines', name="Austin Temperature"),
    row=2, col=2
)

# Update layout to include a title and adjust spacing
fig2.update_layout(
    title_text="Weather Data Comparison: Seattle vs Austin",
    height=800,
    showlegend=True
)

# Show the figure
fig2.show()


print("="*70)

# Create a figure with 2 rows, 1 column, and shared y-axis
fig3 = make_subplots(rows=2, cols=1, shared_yaxes=True, subplot_titles=[
    "Seattle Precipitation",
    "Austin Precipitation"
])

# Add Seattle precipitation data to the top subplot
fig3.add_trace(go.Scatter(
    x=seattle_weather["MONTH"],
    y=seattle_weather["MLY-PRCP-NORMAL"],
    line=dict(color='blue'),
    mode='lines',
    name='Seattle Normal'
), row=1, col=1)

fig3.add_trace(go.Scatter(
    x=seattle_weather["MONTH"],
    y=seattle_weather["MLY-PRCP-25PCTL"],
    line=dict(color='blue', dash='dash'),
    mode='lines',
    name='Seattle 25th Percentile'
), row=1, col=1)

fig3.add_trace(go.Scatter(
    x=seattle_weather["MONTH"],
    y=seattle_weather["MLY-PRCP-75PCTL"],
    line=dict(color='blue', dash='dash'),
    mode='lines',
    name='Seattle 75th Percentile'
), row=1, col=1)

# Add Austin precipitation data to the bottom subplot
fig3.add_trace(go.Scatter(
    x=austin_weather["MONTH"],
    y=austin_weather["MLY-PRCP-NORMAL"],
    line=dict(color='red'),
    mode='lines',
    name='Austin Normal'
), row=2, col=1)

fig3.add_trace(go.Scatter(
    x=austin_weather["MONTH"],
    y=austin_weather["MLY-PRCP-25PCTL"],
    line=dict(color='red', dash='dash'),
    mode='lines',
    name='Austin 25th Percentile'
), row=2, col=1)

fig3.add_trace(go.Scatter(
    x=austin_weather["MONTH"],
    y=austin_weather["MLY-PRCP-75PCTL"],
    line=dict(color='red', dash='dash'),
    mode='lines',
    name='Austin 75th Percentile'
), row=2, col=1)

# Update layout for titles, axis labels, and spacing
fig3.update_layout(
    title="Seattle and Austin Precipitation Data",
    xaxis_title="Month",
    yaxis_title="Precipitation (inches)",
    height=600,
    showlegend=True
)

# Show the figure
fig3.show()


print("="*70)

# Create subplot grid with custom specs
fig4 = make_subplots(
    rows=2,
    cols=2,
    specs=[
        [{"type": "pie", "rowspan": 2}, {"type": "scatter"}],
        [None, {"type": "bar"}]
    ],
    column_widths=[0.4, 0.6],  # Adjust column widths
    vertical_spacing=0.1
)

# Add pie chart spanning two rows (left column)
fig4.add_trace(
    go.Pie(
        labels=["A", "B", "C", "D"],
        values=[20, 30, 15, 35],
        name="Pie Chart"
    ),
    row=1, col=1
)

# Add scatter plot (top right)
fig4.add_trace(
    go.Scatter(
        x=[1, 2, 3, 4],
        y=[10, 11, 12, 13],
        mode="lines+markers",
        name="Scatter Plot"
    ),
    row=1, col=2
)

# Add bar chart (bottom right)
fig4.add_trace(
    go.Bar(
        x=["X", "Y", "Z"],
        y=[5, 3, 7],
        name="Bar Chart"
    ),
    row=2, col=2
)

# Update layout
fig4.update_layout(
    title_text="2x2 Subplot with Spanning Pie Chart",
    height=600,
    showlegend=False
)

fig4.show()




print("="*70)

df_election = px.data.election().head(5)
print("\nElection Data:")
print(df_election)

specs = [[{"type": "Pie"}, {"type": "Pie"}, {"type": "Pie"}]]

fig5 = make_subplots(rows=1, cols=3,
                     specs=specs,
                     subplot_titles=tuple(['Joly Results', 'Coderre Results', 'Bergeron Results'])
                    )

joly_pie = go.Pie(values=df_election['Joly'],
                   labels=df_election['district'])

coderre_pie = go.Pie(values=df_election['Coderre'],
                      labels=df_election['district'])

bergeron_pie = go.Pie(values=df_election['Bergeron'],
                       labels=df_election['district'])

fig5.add_trace(joly_pie, 1, 1)
fig5.add_trace(coderre_pie, 1, 2)
fig5.add_trace(bergeron_pie, 1, 3)

fig5.update_layout(title="Joly vs Coderre vs Bergeron")
fig5.show()




print("="*70)

# Transform the dataframe
new_df = df_election.copy()
new_df = new_df.set_index('district')
new_df = new_df.T
new_df = new_df.iloc[:3]

print("\nTransformed Data:")
print(new_df)

specs = [[{"type": "Pie"}, {"type": "Pie"}, {"type": "Pie"}],
         [{"type": "Pie", "colspan": 2}, None, {"type": "Pie"}]]

fig6 = make_subplots(rows=2, cols=3,
                     specs=specs,
                     subplot_titles=('101-Bois-de-Liesse', '102-Cap-Saint-Jacques',
                                    '11-Sault-au-Récollet', '111-Mile-End', '112-DeLorimier')
                    )

pie_district1 = go.Pie(values=new_df['101-Bois-de-Liesse'],
                       labels=new_df.index)

pie_district2 = go.Pie(values=new_df['102-Cap-Saint-Jacques'],
                       labels=new_df.index)

pie_district3 = go.Pie(values=new_df['11-Sault-au-Récollet'],
                       labels=new_df.index)

pie_district4 = go.Pie(values=new_df['111-Mile-End'],
                       labels=new_df.index)

pie_district5 = go.Pie(values=new_df['112-DeLorimier'],
                       labels=new_df.index)

fig6.add_trace(pie_district1, 1, 1)
fig6.add_trace(pie_district2, 1, 2)
fig6.add_trace(pie_district3, 1, 3)
fig6.add_trace(pie_district4, 2, 1)
fig6.add_trace(pie_district5, 2, 3)

fig6.update_layout(title="Elections Results",
                   height=800,
                   title_x=0.5,
                   title_y=0.95
                  )
fig6.show()



print("="*70)

df = px.data.medals_wide()
print("\nMedals Data:")
print(df)

fig7 = go.Figure()

gold_trace = go.Bar(
    x=df['nation'],
    y=df['gold'],
    marker_color='gold',
    name="gold"
)

silver_trace = go.Bar(
    x=df['nation'],
    y=df['silver'],
    marker_color='grey',
    name="silver"
)

bronze_trace = go.Bar(
    x=df['nation'],
    y=df['bronze'],
    marker_color='red',
    name="bronze"
)

fig7.add_traces([gold_trace, silver_trace, bronze_trace])

# Update layout
fig7.update_layout(
    title='The 2020 Olympic Short Track Speed Skating Medals Results',
    showlegend=True,
    height=600,
    xaxis_title="Nation",
    yaxis_title="Number of Medals"
)

# Show the figure
fig7.show()


print("="*70)
```
